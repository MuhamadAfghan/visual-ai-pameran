# Data Prep — Using Phone (detektor smartphone v2)

Bangun detektor HP dari `raw/using_phone` (Roboflow `using-phone`, segmentasi, kelas face+phone). Kita ambil **kelas phone saja** (poly->bbox) sebagai pengganti detektor `smartphone` lama yang dilatih close-up produk.

**Kenapa lebih baik:** HP berukuran realistis (median ~1% layar, bukan 17%), dan frame berisi orang/wajah → frame tanpa HP jadi **background hard-negative** sehingga model belajar TIDAK menyebut wajah/anak/celana sebagai HP.

Data dibangun oleh skrip generator (nama file pendek anti-MAX_PATH). Notebook ini untuk dokumentasi & profil. Output: `processed/using_phone/` (`nc:1`, `smartphone`).
